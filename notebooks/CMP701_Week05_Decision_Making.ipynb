## 🧠 Short Concept Recap (From Lecture)

Decision making allows a program to **choose between alternative actions**.

In real life:
- If it rains → carry umbrella  
- If score ≥ 50 → pass  

In Python, decisions are made using **Boolean expressions** that evaluate to `True` or `False`.



## 🧪 Lab 5.1: Boolean Expressions

Boolean expressions return either `True` or `False`.


```python

# Examples of Boolean expressions
print(5 > 3)        # Greater than
print(10 == 10)     # Equality check
print(7 != 2)       # Not equal
print(4 <= 6)       # Less than or equal
```


## 🧪 Lab 5.2: The if Statement

The `if` statement executes code **only when a condition is true**.


```python

# Input student score
score = int(input("Enter score: "))

# Check pass condition
if score >= 50:
    print("Passed")
```


## 🧪 Lab 5.3: if–else Statement

The `if–else` statement provides **two alternative paths**.


```python

# Input student score
score = int(input("Enter score: "))

# Determine pass or fail
if score >= 50:
    print("Passed")
else:
    print("Failed")
```


## 🧪 Lab 5.4: elif Statement (Multiple Conditions)

Used when **more than two outcomes** are possible.


```python

# Input student score
score = int(input("Enter score: "))

# Determine grade
if score >= 70:
    print("Grade A")
elif score >= 60:
    print("Grade B")
elif score >= 50:
    print("Grade C")
else:
    print("Fail")
```


## 🧠 Execution Flow (Think Before You Run)

Only **one block** of an if–elif–else structure executes.



## 🧪 Lab 5.5: Nested if Statement

An `if` statement inside another `if` statement.


```python

# Input age and income
age = int(input("Enter age: "))
income = float(input("Enter monthly income: "))

# Loan eligibility check
if age >= 21:
    if income >= 50000:
        print("Eligible for loan")
    else:
        print("Income too low")
else:
    print("Age not eligible")
```

## 🧪 Lab 5.6: Grading Scale as a Reusable Rule
The grading rule below follows the **postgraduate scale used in this course** (A 70–100, B 60–69, C 50–59, F below 50).
*Confirm the boundaries against the current OOU postgraduate regulations before using them for marking.*

```python
def letter_grade(score):
    if not 0 <= score <= 100:       # chained comparison from Week 4
        return "Invalid score"
    if score >= 70:
        return "A"
    elif score >= 60:
        return "B"
    elif score >= 50:
        return "C"
    return "F"

for s in [100, 70, 69, 50, 49, 0, 101, -5]:
    print(s, "->", letter_grade(s))
```


## ⚠️ Common Errors to Avoid
- Using `=` instead of `==`
- Incorrect indentation
- Overlapping elif conditions
- Forgetting the final else
