## 🧠 Short Concept Recap (From Lecture)

An **operator** tells Python to perform an operation.  
An **expression** is a combination of operators and operands that produces a value.

Example:
```
5 + 3 * 2
```
This is an expression containing operators `+` and `*`.



## 🧪 Lab 4.1: Arithmetic Operators

Arithmetic operators are used for mathematical calculations.


````python

# Addition
print(5 + 3)

# Subtraction
print(10 - 4)

# Multiplication
print(6 * 7)

# Division (always returns float)
print(10 / 4)

# Modulus (remainder)
print(10 % 4)

# Exponentiation
print(2 ** 3)

# Floor division
print(10 // 4)
````


### 🔍 Explanation
- `/` gives decimal result
- `//` removes decimal part
- `%` gives the remainder



## 🧪 Lab 4.2: Relational (Comparison) Operators

Relational operators compare values and return **True** or **False**.


````python

# Greater than
print(10 > 5)

# Less than
print(3 < 2)

# Equal to
print(5 == 5)

# Not equal to
print(5 != 3)

# Greater than or equal to
print(7 >= 7)

# Less than or equal to
print(4 <= 6)
````


## 🧪 Lab 4.3: Logical Operators

Logical operators combine multiple conditions.


````python

# Logical AND
print(True and False)

# Logical OR
print(True or False)

# Logical NOT
print(not True)
````


## 🧠 Truth Table Reminder
| A | B | A and B | A or B |
|---|---|---------|--------|
| T | T | T | T |
| T | F | F | T |

| F | F | F | F |



## 🧪 Lab 4.4: Operator Precedence

Python evaluates expressions based on precedence rules.


````python

# Operator precedence example
result = 10 + 5 * 2
print(result)

# Using parentheses to change order
result = (10 + 5) * 2
print(result)
````


### 🧠 Dry-Run Explanation
- Multiplication happens before addition
- Parentheses override precedence



## 🧪 Lab 4.5: Real-World Example – Salary Computation


````python

# Input basic salary
basic_salary = float(input("Enter basic salary: "))

# Compute allowance (20% of basic salary)
allowance = 0.2 * basic_salary

# Compute gross salary
gross_salary = basic_salary + allowance

# Display result
print("Gross Salary =", gross_salary)
````

## 🧪 Lab 4.6: Precedence in Depth
From **highest to lowest** precedence (operators on the same line share a level):

| Level | Operators |
|---|---|
| 1 | `( )` |
| 2 | `**` (**right**-associative: `2 ** 3 ** 2 = 2 ** 9`) |
| 3 | unary `-x`, `+x` |
| 4 | `*`, `/`, `//`, `%` (left-associative) |
| 5 | `+`, `-` |
| 6 | comparisons: `<`, `<=`, `>`, `>=`, `==`, `!=` (and `in`, `is`) |
| 7 | `not` |
| 8 | `and` |
| 9 | `or` |

So `not` binds tighter than `and`, which binds tighter than `or`.

````python
print(-2 ** 2)                  # -(2 ** 2) = -4, not 4
print(2 ** 3 ** 2)              # 2 ** (3 ** 2) = 512
print(True or False and False)  # and first -> True or (False and False) -> True
print(not True or True)         # (not True) or True -> True

score = 75
print(0 <= score <= 100)        # chained comparison: both tests in one expression
````

## 🧪 Lab 4.7: More Assignment Operators and Floor Division

````python
x = 17
x //= 5      # x = x // 5
print(x)
x **= 3      # x = x ** 3
print(x)
x %= 5       # x = x % 5
print(x)

print(-7 // 2, -7 % 2)   # floor division rounds toward negative infinity: -4 1
print(divmod(17, 5))     # quotient and remainder together
````
