## 🧠 Short Concept Recap (From Lecture)

In programming, **variables store data** that programs work with.

Python is dynamically typed, meaning:
- You do not declare variable types explicitly
- Python determines the type automatically

In this lab, we focus on:

## 🧪 Lab 3.1: Variables in Python

A variable is a **named memory location** used to store data.


```python

# Assigning values to variables
age = 25              # age stores an integer value
name = "Sakinat"        # name stores a string value
height = 1.65         # height stores a float value
is_student = True     # is_student stores a boolean value

# Display the variable values
print(age)
print(name)
print(height)
print(is_student)
```


### 🔍 Explanation
- `=` is the assignment operator
- Variables can store different types of data
- Python determines the data type automatically



## 🧪 Lab 3.2: Python Data Types
Python supports several built-in data types.


```python

# Integer type
x = 10
print(type(x))

# Float type
y = 3.14
print(type(y))

# String type
z = "Data Science"
print(type(z))

# Boolean type
flag = False
print(type(flag))
```


## 🧪 Lab 3.3: Input and Output

The `input()` function collects data from the user.


```python

# Collecting user input
name = input("Enter your name: ")

# Displaying output
print("Welcome,", name)
```


## 🧪 Lab 3.4: Type Conversion

All input collected using `input()` is **text (string)** by default.


```python

# Input two numbers as strings
a = input("Enter first number: ")
b = input("Enter second number: ")

# Convert inputs to integers
a = int(a)
b = int(b)

# Perform addition
print("Sum =", a + b)
```

## 🧪 Lab 3.5: Keywords and Identifiers
**Keywords** are reserved words. **Identifiers** (names you invent) must start with a letter or underscore,
contain only letters, digits and underscores, and cannot be keywords. Names are case-sensitive.

```python
import keyword

print(keyword.kwlist)            # every reserved word in this Python version

for candidate in ["age", "student_name", "_total", "2age", "student-name", "if", "Age"]:
    ok = candidate.isidentifier() and not keyword.iskeyword(candidate)
    print(f"{candidate!r:16} valid identifier? {ok}")
```

## 🧪 Lab 3.6: Neat Output with f-strings
An **f-string** (formatted string literal) puts values straight into text. The part after `:` controls formatting.

```python
name = "Aisha"
age = 30
average = 66.66666

print(f"Welcome {name}, you are {age} years old.")
print(f"Average score = {average:.2f}")      # 2 decimal places
print(f"Salary: ₦{1250000:,}")               # thousands separator
```

## 🧪 Lab 3.7: What Happens with Bad Input?
`int()` raises a `ValueError` when the text is not a whole number. (We handle such errors properly in Week 13;
for now, just *read* the message — reading error messages is a core skill.)

```python
try:
    int("twenty")
except ValueError as err:
    print("ValueError:", err)
```
