## 🧠 Concept Recap
So far we used a **procedural** style: functions operate on separate data (for example, a dictionary).
**Object-Oriented Programming (OOP)** bundles the data and the functions that work on it into **objects**.
A **class** is the blueprint; an **object** (instance) is one thing built from it.

## 🧪 Lab 14.1: Same Problem, Two Styles

````python
# Procedural: a dictionary plus separate functions
def average_of(record):
    return sum(record["scores"]) / len(record["scores"])

record = {"name": "Aisha", "scores": [70, 65, 80]}
print(f"{record['name']}: {average_of(record):.1f}")
````

````python
# Object-oriented: data and behaviour live together
class Student:
    def __init__(self, name, scores):
        self.name = name              # attributes
        self.scores = list(scores)

    def average(self):                # method
        return sum(self.scores) / len(self.scores)

s1 = Student("Aisha", [70, 65, 80])
print(f"{s1.name}: {s1.average():.1f}")
````

**Key ideas**
- `__init__` is the **constructor**: it runs when an object is created and sets its attributes.
- `self` is the object the method is working on. Python passes it automatically.
- Each object has its **own** attribute values.

| Feature | Procedural | Object-oriented |
|---|---|---|
| Organised around | Functions | Objects (data + behaviour) |
| Data and logic | Separate | Bundled |
| Good for | Small scripts | Larger programs modelling real entities |

## 🧪 Lab 14.2: A More Complete Student Class

````python
class Student:
    """A student with a name, matric number and a list of course scores."""

    def __init__(self, name, matric, scores):
        if not scores:
            raise ValueError("a student needs at least one score")
        if any(not 0 <= s <= 100 for s in scores):
            raise ValueError("scores must be between 0 and 100")
        self.name = name
        self.matric = matric
        self.scores = list(scores)

    def total(self):
        return sum(self.scores)

    def average(self):
        return self.total() / len(self.scores)

    def grade(self):
        avg = self.average()
        if avg >= 70: return "A"
        if avg >= 60: return "B"
        if avg >= 50: return "C"
        return "F"

    def __str__(self):                # controls what print(student) shows
        return f"{self.name} ({self.matric}): average {self.average():.2f}, grade {self.grade()}"

s = Student("Aisha", "CMP/PGD/001", [72, 65, 80])
print(s)

try:
    Student("Bola", "CMP/PGD/002", [101])
except ValueError as err:
    print("Rejected:", err)
````

## 🧪 Lab 14.3: BankAccount with a Custom Exception

````python
class InsufficientFunds(Exception):
    """Raised when a withdrawal exceeds the balance."""


class BankAccount:
    def __init__(self, account_no, balance=0):
        self.account_no = account_no
        self._balance = balance            # leading underscore: "internal, please don't touch directly"

    @property
    def balance(self):                     # read-only access: account.balance
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self._balance += amount

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("withdrawal must be positive")
        if amount > self._balance:
            raise InsufficientFunds(f"balance {self._balance} is less than {amount}")
        self._balance -= amount


acct = BankAccount("0123456789", 50_000)
acct.deposit(20_000)
acct.withdraw(30_000)
print("Balance:", acct.balance)

try:
    acct.withdraw(1_000_000)
except InsufficientFunds as err:
    print("InsufficientFunds:", err)
````

## 🧪 Lab 14.4: Many Objects, Sorting and Saving to CSV

````python
import csv

students = [
    Student("Aisha", "CMP/PGD/001", [72, 65, 80]),
    Student("Bola", "CMP/PGD/002", [55, 48, 61]),
    Student("Chinedu", "CMP/PGD/003", [90, 85, 88]),
]

best = max(students, key=Student.average)             # method used as the sort key
print("Best:", best)

for st in sorted(students, key=Student.average, reverse=True):
    print(st)

with open("class_results.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["matric", "name", "average", "grade"])
    for st in students:
        writer.writerow([st.matric, st.name, f"{st.average():.2f}", st.grade()])

print(open("class_results.csv", encoding="utf-8").read())
````
