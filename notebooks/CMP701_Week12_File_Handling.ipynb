## 🧪 Lab 12.1: Writing to a Text File
Always pass `encoding="utf-8"`. Without it, Python uses the operating system's default (often not UTF-8 on Windows),
and names such as *Adéwálé* or *Ọbáfẹ́mi* can be corrupted or raise errors.

```python
with open("students.txt", "w", encoding="utf-8") as file:
    file.write("Aisha,78\n")
    file.write("Bola,85\n")
    file.write("Chinedu,90\n")
print("Data written successfully")
```

## 🧪 Lab 12.2: Reading a Whole File

```python
with open("students.txt", "r", encoding="utf-8") as file:
    content = file.read()
print(content)
```

## 🧪 Lab 12.3: Reading Line by Line

```python
with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())          # strip() removes the newline character
```

## 🧪 Lab 12.4: File Modes
| Mode | Meaning | If the file exists | If it does not exist |
|---|---|---|---|
| `r` | read | read it | `FileNotFoundError` |
| `w` | write | **erases** existing content | creates it |
| `a` | append | adds to the end | creates it |
| `x` | exclusive create | `FileExistsError` | creates it |

```python
with open("students.txt", "a", encoding="utf-8") as file:
    file.write("Zainab,88\n")
print("New record added")

try:
    with open("students.txt", "x", encoding="utf-8") as file:   # file already exists
        file.write("this will not be written")
except FileExistsError as err:
    print("FileExistsError:", err)
```

## 🧪 Lab 12.5: Handling File Errors

```python
try:
    with open("nonexistent.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("File not found. Please check the filename.")
```

## 🧪 Lab 12.6: Parsing Records Safely
Splitting on commas by hand breaks on blank lines and on fields that contain commas. Skip blanks and use `int()` inside `try`.

```python
total = 0
count = 0
with open("students.txt", "r", encoding="utf-8") as file:
    for line in file:
        line = line.strip()
        if not line:                     # skip blank lines
            continue
        name, score_text = line.split(",")
        try:
            score = int(score_text)
        except ValueError:
            print("Skipping bad score for", name)
            continue
        total += score
        count += 1

print(f"{count} records, average = {total / count:.2f}")
```

## 🧪 Lab 12.7: CSV Files with the `csv` Module
CSV (Comma-Separated Values) stores tabular data. The `csv` module handles quoting, commas inside fields, and blank lines.

```python
import csv

rows = [
    {"name": "Aisha", "score": 78},
    {"name": "Adéwálé, Tunde", "score": 85},      # a comma inside the name is handled correctly
    {"name": "Ọbáfẹ́mi", "score": 90},
]

with open("scores.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["name", "score"])
    writer.writeheader()
    writer.writerows(rows)

with open("scores.csv", "r", newline="", encoding="utf-8") as f:
    data = list(csv.DictReader(f))

for row in data:
    print(row["name"], row["score"])

scores = [int(row["score"]) for row in data]
print(f"Average = {sum(scores) / len(scores):.2f}")
```

## 🧪 Lab 12.8: Simple File-Based Student Record System

```python
import csv, os

FILENAME = "records.csv"

def add_record(name, score):
    new_file = not os.path.exists(FILENAME)
    with open(FILENAME, "a", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        if new_file:
            writer.writerow(["name", "score"])
        writer.writerow([name, score])

def show_records():
    with open(FILENAME, "r", newline="", encoding="utf-8") as f:
        for row in csv.DictReader(f):
            print(f"{row['name']:<20}{row['score']:>5}")

name = input("Enter student name: ")
score = input("Enter score: ")
add_record(name, score)
print("Student Records:")
show_records()
```
