## 🧠 Short Concept Recap (From Lecture)

In **Week 06**, we used the `while` loop for condition-controlled repetition.

In **Week 07**, we use the `for` loop when:
- The number of repetitions is known
- We are iterating over a sequence (numbers, list, string)

Think of the `for` loop as:
> “For each value in this sequence, do something.”



## 🔁 while vs for (Quick Comparison)

| Feature | while loop | for loop |
|------|-----------|---------|
| Control | Condition-based | Count/sequence-based |
| Best for | Unknown repetitions | Known repetitions |
| Variable update | Manual | Automatic |



## 🧪 Lab 7.1: Basic for Loop


```python

# Print numbers from 1 to 5 using for loop
for i in range(1, 6):
    print(i)
```


### 🔍 Explanation
- `range(1, 6)` generates numbers from 1 to 5
- The loop variable `i` changes automatically



## 🧪 Lab 7.2: Understanding range()

The range() function can take 1, 2, or 3 arguments.


```python

# range(stop)
print(list(range(5)))

# range(start, stop)
print(list(range(1, 6)))

# range(start, stop, step)
print(list(range(1, 11, 2)))
```


## 🧪 Lab 7.3: Multiplication Table


```python

# Input a number
num = int(input("Enter a number: "))

# Generate multiplication table
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```


## 🧪 Lab 7.4: break Statement

The break statement stops the loop immediately.


```python

# Stop loop when i reaches 5
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```


## 🧪 Lab 7.5: continue Statement

The continue statement skips the current iteration.


```python

# Skip printing number 3
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```


## 🧪 Lab 7.6: Nested Loops

A loop inside another loop.


```python

# Print a square pattern using nested loops
for i in range(1, 4):
    for j in range(1, 4):
        print("*", end=" ")
    print()  # New line after inner loop
```


## 🧠 Dry-Run Thinking

Outer loop controls rows  
Inner loop controls columns


## 🧪 Lab 7.7: Iterating over Strings and Lists
A `for` loop works on any sequence, not only `range()`.

```python
for letter in "PYTHON":
    print(letter, end=" ")
print()

courses = ["CMP701", "CMP702", "CMP703"]
for index, course in enumerate(courses, start=1):
    print(index, course)
```

## 🧪 Lab 7.8: A Simple Simulation (reproducible)
Fixing the random **seed** makes a simulation give the same result every run — essential for reproducible work.

```python
import random

random.seed(42)                       # reproducibility
rolls = [random.randint(1, 6) for _ in range(12)]
print("Rolls:", rolls)

sixes = 0
for r in rolls:
    if r == 6:
        sixes += 1
print("Sixes:", sixes, "out of", len(rolls))
```
