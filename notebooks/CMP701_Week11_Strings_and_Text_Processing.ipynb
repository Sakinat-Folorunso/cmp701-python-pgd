## 🧠 Concept Recap: What is a String?

A **string** is a sequence of characters enclosed in quotation marks.
Strings are used to store and process text data such as:
- Names
- Sentences
- Passwords
- Documents

In Python, strings are **immutable**.



## 🧪 Lab 11.1: Creating and Displaying Strings


## 🧪 Lab 11.2: String Indexing

Each character in a string has an index starting from 0.


```python

text = "PYTHON"

# Access characters using index
print(text[0])   # First character
print(text[3])   # Fourth character
print(text[-1])  # Last character
```


## 🧪 Lab 11.3: String Slicing

Slicing extracts part of a string.


```python

word = "COMPUTER"

print(word[0:4])   # COMP
print(word[3:])    # PUTER
print(word[:5])    # COMPU
```


## 🧪 Lab 11.5: Common String Methods


```python

sentence = "  Welcome to CMP701 Programming  "

print(sentence.lower())     # Convert to lowercase
print(sentence.upper())     # Convert to uppercase
print(sentence.strip())     # Remove leading and trailing spaces
print(sentence.replace("Programming", "Python"))
print(len(sentence))        # Length of string
```


## 🧪 Lab 11.6: Splitting Strings


```python

sentence = "Data Science and Artificial Intelligence"

# Split sentence into words
words = sentence.split()

print(words)
print("Number of words:", len(words))
```

## 🧪 Lab 11.9: Formatting Strings and More Methods

```python
name, score = "Aisha", 78.456

print(f"{name:<10}|{score:>8.2f}|")     # left/right alignment and 2 decimals
print("data science".title())            # Data Science
print("-".join(["CMP", "701"]))          # CMP-701
print("banana".count("an"))              # 2
print("computer".find("p"))              # 3  (find returns -1 if absent)
print("Python".startswith("Py"), "Python".endswith("on"))
```

## 🧪 Lab 11.10: Simple Text Analyzer

```python
text = "Data science helps Nigerian researchers solve local problems"

words = text.split()
longest = max(words, key=len)

print(f"Characters (with spaces): {len(text)}")
print(f"Characters (without spaces): {len(text.replace(' ', ''))}")
print(f"Words: {len(words)}")
print(f"Longest word: {longest} ({len(longest)} letters)")
```

## 🧪 Lab 11.11: Strings are Unicode — a Yorùbá Example
Python 3 strings hold Unicode text, but an accented letter can be stored as **one** code point or as a base letter **plus**
combining marks. The visible text looks identical; `len()` does not. This matters for NLP on Yorùbá, Igbo and other
tonal languages.

```python
import unicodedata

word = "Ọlọ́run"
composed = unicodedata.normalize("NFC", word)
decomposed = unicodedata.normalize("NFD", word)

print(composed, decomposed)
print(len(composed), len(decomposed))     # 7 vs 9 code points
print(composed == decomposed)             # False, although they look the same
print(unicodedata.normalize("NFC", decomposed) == composed)   # True after normalising
```
