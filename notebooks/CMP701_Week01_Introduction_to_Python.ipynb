## 🧪 Lab 1.4: What Really Happens When Python Runs Your Code?
Python is usually called an *interpreted* language, but the full picture is a little richer.
The standard implementation, **CPython**, first **compiles** your source code into **bytecode**,
and then the **Python virtual machine** executes that bytecode instruction by instruction.

So Python is neither "compiled to machine code ahead of time" (like a typical C program) nor
purely "read line by line". The cell below shows the bytecode of a tiny function.

```python
import dis

def add(a, b):
    return a + b

dis.dis(add)   # the exact instructions differ between Python versions
```

**Think about it:** if bytecode is produced before execution starts, why does a syntax error on the
*last* line of a script still stop the program *before* the first line runs?
