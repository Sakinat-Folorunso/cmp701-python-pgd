## 🧠 Concept Recap: From Lists to Records
Lists store values by **position**. Real-world data often has **meaningful labels**.

Example: student record → name, matric, department, score. This motivates **dictionaries**.

## 🧪 Lab 9.1: Creating a Dictionary
A dictionary stores data as **key : value** pairs. Keys must be unique and hashable (strings, numbers, tuples).

> **Ordering.** Since Python 3.7, dictionaries **preserve insertion order**. They are not indexed by position, but they are *not* "unordered".

```python
student = {
    "name": "Aisha",
    "matric": "CMP/PGD/001",
    "department": "Computer Science",
    "score": 78,
}
print(student)
print(list(student))      # keys come back in the order they were inserted
```

## 🧪 Lab 9.2: Accessing Dictionary Values

```python
print(student["name"])
print(student.get("score"))            # safer access
print(student.get("email"))            # None instead of an error
print(student.get("email", "n/a"))     # custom default

try:
    print(student["email"])
except KeyError as err:
    print("KeyError:", err)
```

## 🧪 Lab 9.3: Updating, Adding and Removing

```python
student["score"] = 85            # update
student["level"] = "PGD"         # add
removed = student.pop("department")
print("removed:", removed)
print(student)
print("level" in student, "email" in student)   # membership tests check KEYS
```

## 🧪 Lab 9.4: Looping Through Dictionaries

```python
for key, value in student.items():
    print(f"{key:>8} : {value}")

print(list(student.keys()))
print(list(student.values()))
```

## 🧪 Lab 9.5: A List of Dictionaries (Student Record System)

```python
students = []
for i in range(2):
    record = {}
    record["name"] = input("Enter name: ")
    record["matric"] = input("Enter matric: ")
    record["score"] = int(input("Enter score: "))
    students.append(record)

for s in students:
    print(s)
```

## 🧪 Lab 9.6: Frequency Counting with a Dictionary
Counting is a dictionary job (a set cannot count). The pattern is *get the current count, or 0, then add one*.

```python
courses_registered = ["CMP701", "CMP702", "CMP701", "CMP703", "CMP701", "CMP702"]

counts = {}
for course in courses_registered:
    counts[course] = counts.get(course, 0) + 1
print(counts)

# The standard-library shortcut
from collections import Counter
print(Counter(courses_registered))
print(Counter(courses_registered).most_common(1))
```

## 🧪 Lab 9.7: Introduction to Sets
A set stores **unique** elements only. Order is **not** guaranteed.

```python
courses = {"CMP701", "CMP702", "CMP701", "CMP703"}
print(len(courses), sorted(courses))     # use sorted() when you need a predictable display order

print(type({}))        # <class 'dict'>  -> {} is an EMPTY DICTIONARY
print(type(set()))     # <class 'set'>   -> use set() for an empty set
```

## 🧪 Lab 9.8: Set Operations

```python
A = {1, 2, 3}
B = {3, 4, 5}

print("Union:", sorted(A | B))
print("Intersection:", sorted(A & B))
print("Difference A-B:", sorted(A - B))
print("Symmetric difference:", sorted(A ^ B))
print("Subset?", {1, 2} <= A)
```

## 🧠 Exam Insight: Data Structure Comparison

| Structure | Stores | Ordered? | Duplicates | Mutable |
|---|---|---|---|---|
| List | Values | Yes (by position) | Yes | Yes |
| Tuple | Values | Yes (by position) | Yes | No |
| Dictionary | Key–value pairs | Insertion-ordered (Python ≥ 3.7); looked up by key | Keys unique | Yes |
| Set | Values | No guaranteed order | No | Yes |
