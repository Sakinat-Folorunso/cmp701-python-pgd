## 🧠 Concept Recap: Why Functions?

As programs grow larger, writing all code in one place becomes:
- Hard to read
- Difficult to debug
- Impossible to reuse efficiently

Functions allow us to **break a program into smaller, reusable blocks**.


## 🧪 Lab 10.6: Menu-Driven Program


```python

# Menu-driven calculator

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

while True:
    print("1. Add")
    print("2. Subtract")
    print("3. Exit")
    
    choice = input("Choose option: ")
    
    if choice == "1":
        x = int(input("Enter first number: "))
        y = int(input("Enter second number: "))
        print("Result:", add(x, y))
    elif choice == "2":
        x = int(input("Enter first number: "))
        y = int(input("Enter second number: "))
        print("Result:", subtract(x, y))
    elif choice == "3":
        break
    else:
        print("Invalid choice")
```

## 🧪 Lab 10.7: Safer Division (guard clauses)

```python
def divide(a, b):
    """Return a / b, or None when b is zero."""
    if b == 0:
        return None
    return a / b

print(divide(10, 4))
print(divide(10, 0))
```

## 🧪 Lab 10.8: Default Arguments, Keyword Arguments and Docstrings
- **Default arguments** must come *after* non-default ones.
- **Keyword arguments** let you pass values by name, in any order.
- A **docstring** (the first string in a function) documents it and is shown by `help()`.

```python
def greet(name, message="Welcome"):
    """Print a greeting for `name` using `message`."""
    print(message, name)

greet("Aisha")
greet("Bola", "Good morning")
greet(message="Hello", name="Chinedu")   # keyword arguments

help(greet)
```

**Pitfall — mutable default arguments.** A default list is created *once*, when the function is defined, and then shared by every call.

```python
def add_score_bad(score, scores=[]):      # DON'T do this
    scores.append(score)
    return scores

print(add_score_bad(60))
print(add_score_bad(70))     # surprise: [60, 70]

def add_score_good(score, scores=None):   # do this instead
    if scores is None:
        scores = []
    scores.append(score)
    return scores

print(add_score_good(60))
print(add_score_good(70))    # [70]
```

## 🧪 Lab 10.9: Scope, `global`, and a Safe Recursion Example

```python
counter = 0

def increment_bad():
    counter = counter + 1        # UnboundLocalError: Python treats `counter` as local here

try:
    increment_bad()
except UnboundLocalError as err:
    print("UnboundLocalError:", err)

def increment():
    global counter               # works, but prefer returning values over mutating globals
    counter += 1

increment()
print(counter)
```

```python
def factorial(n):
    """Return n! for a non-negative integer n."""
    if n < 0:
        raise ValueError("factorial is undefined for negative numbers")
    if n <= 1:                   # base case covers both 0 and 1
        return 1
    return n * factorial(n - 1)

print([factorial(k) for k in range(6)])   # [1, 1, 2, 6, 24, 120]
```
