## 🧪 Lab 8.1: Creating a List
A list is an ordered, **mutable** collection enclosed in square brackets `[]`.

```python
scores = [65, 70, 55, 80, 75]
print(scores)
print(type(scores), len(scores))
```

## 🧪 Lab 8.2: List Indexing
Indexing starts at **0**. Negative indices count from the end.

```python
print(scores[0])    # first
print(scores[2])    # third
print(scores[-1])   # last
print(scores[-2])   # second from last
```

## 🧪 Lab 8.3: List Slicing
`list[start:stop:step]` — the `stop` index is **not** included.

```python
print(scores[1:4])    # [70, 55, 80]
print(scores[:3])     # first three
print(scores[2:])     # from index 2 to the end
print(scores[::2])    # every second element
print(scores[::-1])   # the list reversed (a new list)
```

## 🧪 Lab 8.4: Common List Methods (watch the list change)

```python
scores = [65, 70, 55, 80, 75]

scores.append(90);      print("append(90)    ->", scores)
scores.insert(2, 60);   print("insert(2, 60) ->", scores)
scores.remove(55);      print("remove(55)    ->", scores)
removed = scores.pop(1); print(f"pop(1) removed {removed} ->", scores)
```

## 🧠 Memory View (Think Like the Computer)
The cell below *prints* the index/value map from the **actual** list, so the picture can never disagree with the code.

```python
print("Index:", *[f"{i:>4}" for i in range(len(scores))])
print("Value:", *[f"{v:>4}" for v in scores])
```

## 🧪 Lab 8.5: More Useful Methods

```python
marks = [72, 45, 88, 61, 45]

print(marks.count(45))        # how many times 45 occurs
print(marks.index(88))        # position of first 88
print(88 in marks)            # membership test
marks.extend([50, 90])        # add several items
print(marks)

marks.sort()                  # sorts in place, returns None
print(marks)
marks.sort(reverse=True)
print(marks)

print(sorted([3, 1, 2]))      # sorted() returns a NEW list and leaves the original alone
print(sum(marks), min(marks), max(marks))
```

## 🧪 Lab 8.6: Aliasing vs Copying (a classic source of bugs)
Assigning a list to another name does **not** copy it — both names refer to the **same** list.

```python
a = [1, 2, 3]
b = a            # alias: same list object
b.append(99)
print(a, b, a is b)

c = a.copy()     # real copy (also: list(a) or a[:])
c.append(100)
print(a, c, a is c)
```

## 🧪 Lab 8.7: Processing List Data (Student Scores)

```python
scores = []
for i in range(5):
    scores.append(int(input(f"Enter score {i + 1}: ")))

print("Scores:", scores)
print("Highest score:", max(scores))
print("Lowest score:", min(scores))
print(f"Average score: {sum(scores) / len(scores):.2f}")
```

## 🧪 Lab 8.8: Introduction to Tuples
A tuple is ordered and **immutable**. It is ideal for fixed records and can be *unpacked*.

```python
student = ("Aisha", "CMP/PGD/001", 78)

print(student[0], student[2])
name, matric, score = student        # tuple unpacking
print(f"{name} ({matric}) scored {score}")

try:
    student[2] = 90
except TypeError as err:
    print("TypeError:", err)

# Tuples are hashable, so they can be dictionary keys (lists cannot)
locations = {(6.62, 3.97): "Ago-Iwoye"}
print(locations[(6.62, 3.97)])
```

## 🧠 Lists vs Tuples (Exam Favourite)

| Feature | List | Tuple |
|---|---|---|
| Brackets | `[ ]` | `( )` |
| Mutable | Yes | No |
| Hashable (usable as dict key) | No | Yes (if all items are hashable) |
| Typical use | Changing collections of similar items | Fixed records of mixed items |
| Speed / memory | Slightly more overhead | Slightly leaner |
