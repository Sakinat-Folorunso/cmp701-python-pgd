## 🧠 Short Concept Recap (From Lecture)

Many problems require **repeating actions**:
- Counting numbers
- Summing values
- Repeating input until a condition is met

In Python, repetition is achieved using **loops**.
This week focuses on the **while loop**, which repeats **as long as a condition is true**.



## 🔁 The while Loop Structure

General syntax:
```python
while condition:
    statement(s)
```
The loop continues **while the condition is True**.



## 🧪 Lab 6.1: Counting Numbers Using while Loop


````python

# Initialize a counter variable
i = 1

# Loop continues while i is less than or equal to 5
while i <= 5:
    print(i)      # Display current value of i
    i += 1        # Increment i by 1
````


### 🔍 Explanation
- `i` is the loop control variable
- The condition is checked before each iteration
- The loop stops when the condition becomes False



## 🧪 Lab 6.2: Sum of First 10 Natural Numbers


````python

# Initialize sum and counter
total = 0
i = 1

# Loop to compute sum
while i <= 10:
    total += i    # Add current i to total
    i += 1        # Move to next number

# Display result
print("Sum =", total)
````

## 🧠 Dry-Run Table (Very Important)

For the loop above:

| Iteration | i (before) | Condition `i <= 10` | total (after) | i (after) |
|---|---|---|---|---|
| 1 | 1 | True | 1 | 2 |
| 2 | 2 | True | 3 | 3 |
| 3 | 3 | True | 6 | 4 |
| 4 | 4 | True | 10 | 5 |
| 5 | 5 | True | 15 | 6 |
| 6 | 6 | True | 21 | 7 |
| 7 | 7 | True | 28 | 8 |
| 8 | 8 | True | 36 | 9 |
| 9 | 9 | True | 45 | 10 |
| 10 | 10 | True | 55 | 11 |
| — | 11 | False | loop ends | — |

> Note: we named the accumulator `total`, **not** `sum`. Naming a variable `sum` hides Python's built-in
> `sum()` function and causes `TypeError: 'int' object is not callable` later in the same session.


## 🧪 Lab 6.3: Password Retry System

This program keeps asking for a password until the correct one is entered.


````python

# Correct password
password = "admin"

# Initial user input
user_input = ""

# Repeat until correct password is entered
while user_input != password:
    user_input = input("Enter password: ")

print("Access granted")
````

## 🧪 Lab 6.4: The Input-Validation Loop (a pattern you will reuse all semester)
Keep asking until the user gives acceptable input. This pattern appears again in Weeks 13–14 and the mini-project.

````python
while True:
    text = input("Enter a score (0-100): ")
    if text.isdigit() and 0 <= int(text) <= 100:
        score = int(text)
        break                      # leave the loop only when the input is valid
    print("Invalid input. Try again.")

print("Accepted score:", score)
````
